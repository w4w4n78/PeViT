```python
import os
import glob
import scipy.io
from torcheeg import transforms
from HBUEDDataset import HBUEDDataset
from torcheeg.io import EEGSignalIO, MetaInfoIO

window = 20
c_size = 200
```

```python
dataset = HBUEDDataset(
    io_path=f'../HBUED/tmp_out/PeViT',
    root_path='../../Dataset/HBUED/original',
    offline_transform=transforms.Compose([
        transforms.MeanStdNormalize()
    ]),
    online_transform=transforms.Compose([
        transforms.ToTensor()
    ]),
    label_transform=transforms.Compose([
        transforms.Select(['arousal', 'valence']),
        transforms.Binary(5.0)
    ]),
    overlap=c_size - window,
    chunk_size=c_size
)

print(len(dataset))
x, y = dataset[0]
print(x.shape, y)
```
